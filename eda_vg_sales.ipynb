```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import plotly.express as px
import plotly.io as pio
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

```python
df_sales_ind = pd.read_csv('sales_ind_clean.csv', index_col=False, parse_dates=["release_date"])
df_sales_series = pd.read_csv('sales_series_clean.csv', index_col=False, parse_dates=["release_date"])

sns.set_style("whitegrid")
pio.templates.default = "seaborn"
```

```python
# 1 - Evolución del mercado a lo largo del tiempo

# ¿Cómo han evolucionado las ventas de los videojuegos a lo largo de los años?

sales_per_year = df_sales_ind.groupby('year').agg(
    sales=('units_reported', 'sum'),
    games_released=('units_reported', 'count'),
    sales_mean = ('units_reported', 'mean'),
    sales_median = ('units_reported', 'median')
).reset_index().copy()

# print(sales_per_year)

# Gráfico ventas totales por año de origen

fig = px.line(sales_per_year, x="year", y="sales", markers=True,
              title='Ventas totales según año de lanzamiento',
              hover_data=['games_released'])

fig.update_layout(
    xaxis_title="Año de lanzamiento",
    yaxis_title="Ventas totales (millones de copias)"
)

fig.update_traces(line_color="crimson")

fig.show()


# Gráfico promedio de ventas por juego según año de origen

fig = px.line(sales_per_year, x="year", y="sales_mean", markers=True,
              title='Ventas promedio según año de lanzamiento')

fig.update_layout(
    xaxis_title="Año de lanzamiento",
    yaxis_title="Promedio de ventas"
)

fig.show()


# Gráfico mediana de ventas por juego según año de origen

fig = px.line(sales_per_year, x="year", y="sales_median", markers=True,
              title='Mediana de ventas según año de lanzamiento')

fig.update_layout(
    xaxis_title="Año de lanzamiento",
    yaxis_title="Mediana de ventas"
)

fig.update_traces(line_color="#19D3F3")

fig.show()

# Gráfico cantidad de juegos según año de origen

#fig = px.line(sales_per_year, x="year", y="games_released", markers=True,
 #             title='Cantidad de juegos lanzados por año')
fig = px.histogram(df_sales_ind, x="year", color_discrete_sequence=['#00CC96'] )

fig.update_layout(
    xaxis_title="Año de lanzamiento",
    yaxis_title="Cantidad de juegos"
)

fig.update_traces(
    marker_line_color='black',  # Color of the line between/around bars
    marker_line_width=0.5       # Thickness of the line
)

fig.show()
```

```python
# 1.5 - Estudio menor: Variación de ventas por región en diferentes años de origen

df_sales_region = df_sales_ind[df_sales_ind['total_sales'].notna()]

sales_per_region = df_sales_region.groupby('year').agg(
    sales=('total_sales', 'sum'),
    games_released=('total_sales', 'count'),
    na_sales=('na_sales', 'sum'),
    jp_sales=('jp_sales', 'sum'),
    pal_sales=('pal_sales', 'sum'),
    other_sales=('other_sales', 'sum'),
).reset_index().copy()
sales_per_region = sales_per_region[sales_per_region['year'].between(1998, 2018)]

regional_sales = ['na_sales', 'pal_sales', 'other_sales', 'jp_sales']
important_years = [2008, 2011, 2017]

df_long_region = df_sales_region.melt(
    id_vars=["year", "title"],
    value_vars=regional_sales,
    var_name="region",
    value_name="region_sales"
)

df_long_region = df_long_region.sort_values("region_sales", ascending=False)

# print(df_long_region)

fig = px.bar(sales_per_region, x='year', y=regional_sales)

fig.update_layout(
    title="Ventas por región (año de origen)",
    xaxis_title="Año de lanzamiento",
    yaxis_title="Ventas"
)

fig.show()



for year in important_years:
    df_long_year = df_long_region[df_long_region['year'] == year].copy()
    df_long_year = df_long_year[df_long_year['region_sales'] > 0]
    df_long_year["ranking"] = range(1, len(df_long_year) + 1)
    fig = px.scatter(
    df_long_year,
    x="ranking",
    y="region_sales",
    color="region",
    hover_name="title"
    )
    fig.update_layout(
    title=f"Mayores ventas de juegos año {year}",
    xaxis_title="Ranking",
    yaxis_title="Ventas"
    )
    fig.show()    


```

```python
# 2 - Estudio de ventas por género

sales_by_genre = df_sales_ind.groupby('genre').agg(
    sales=('units_reported', 'sum'),
    games_released=('units_reported', 'count'),
    sales_mean = ('units_reported', 'mean'),
    sales_median = ('units_reported', 'median')
).reset_index().sort_values('sales', ascending=False).copy()

df_long = sales_by_genre.melt(id_vars ='genre', value_vars=['sales_mean', 'sales_median'],
                              var_name='statistic', value_name='value')

# Gráfico ventas totales
fig = px.bar(sales_by_genre, x='genre', y='sales')

fig.update_layout(
    title="Ventas por género",
    xaxis_title="Género",
    yaxis_title="Ventas"
)
fig.update_xaxes(tickangle=60) 
fig.update_traces(marker_color='crimson')

fig.show()


# Gráfico Top 10
top_10_genres = sales_by_genre.iloc[:10]

fig = px.pie(top_10_genres, values='sales', names='genre', title='Top 10 géneros con más ventas totales')
fig.show()


# Gráfico media vs. mediana
fig = px.bar(df_long, x='genre', y='value', color='statistic', barmode='group')
fig.update_layout(
    title="Promedio vs. Mediana de ventas por género",
    xaxis_title="Género",
    yaxis_title="Promedio/Mediana"
)
fig.update_xaxes(tickangle=60) 
fig.show()


# Gráfico de cantidad de juegos por género

fig = px.histogram(df_sales_ind, x="genre", color_discrete_sequence=['#00CC96'] )
fig.update_layout(
    title="Juegos lanzados por género",
    xaxis_title="Género",
    yaxis_title="Cantidad de juegos"
)
fig.update_xaxes(tickangle=60) 
fig.show()


# Gráfico 10% 
sales_top_10_percent = df_sales_ind.iloc[:int(len(df_sales_ind) * 0.10)]

proportion_10 = sales_top_10_percent['genre'].value_counts()
proportion_full = df_sales_ind['genre'].value_counts()
genre_proportion = (proportion_10 / proportion_full).fillna(0).sort_values(ascending=False)

fig = px.bar(x=genre_proportion.index, y=genre_proportion)
fig.update_yaxes(tickformat=".0%")
fig.update_layout(
    title="Presencia de géneros en entregas exitosas",
    xaxis_title="Género",
    yaxis_title="% en mejores vendidos"
)
fig.update_xaxes(tickangle=60) 
fig.show()


# Gráfico de juegos más vendidos de cada género

important_genres = ["Action-Adventure", "Shooter", "Role-Playing", "Sports", "Party", "Platform"]
top_10_unique = sales_top_10_percent.groupby(['genre']).head(2).copy()
top_10_unique = top_10_unique[top_10_unique['genre'].isin(important_genres)]

fig = px.scatter(top_10_unique, y="genre", x="units_reported", color="title")
fig.update_layout(
    title="Títulos más exitosos por género",
    xaxis_title="Ventas",
    yaxis_title="Géneros"
)
fig.update_traces(marker_size=15)
fig.show()
```
